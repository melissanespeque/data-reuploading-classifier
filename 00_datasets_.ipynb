# Datasets:

Implementation of binary classification datasets - from the datasets provided by the paper, and the Iris classification dataset.

```python
## Libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
import sys
import warnings
warnings.filterwarnings('ignore') 
from datasets import circle, plot_circle, diamond, plot_diamond, wavy_lines, plot_wavy
```

## Synthetic Datasets

### Circle

```python
data, (centers, radii) = circle(samples=3000, random_seed=42)
fig = plot_circle(data, centers, radii)
plt.show()
```

```python
#Train/test split
data, (centers, radii) = circle(samples=3000, random_seed=42)
x = np.array([i[0] for i in data])
y = np.array([i[1] for i in data])

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y) #stratify allows to have a balance amount of labels in each split


print("Train/Test datasets")
print(f"Train: {len(x_train)} samples")
print(f"Test: {len(x_test)} samples")
      
```

```python
#Labels
print("Labels:")
print("0 <- Inside the circle")
print("1 <- Outside the circle")
print(np.array([i[1] for i in data]))
```

### Diamond

```python
x_data, y_data, logic = diamond(samples=5000, limit=10, scale=4.0)
fig = plot_diamond(x_data, y_data, logic)
plt.show()
```

```python
#Train/test split
x_data, y_data, pattern_labels = diamond(samples=5000, limit=10, scale=4.0)

x = np.column_stack((x_data, y_data))
y = pattern_labels

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)

print("Train/Test datasets")
print(f"Train: {len(x_train)} samples")
print(f"Train: {len(x_test)} samples")
```

```python
#Labels
print("Labels:")
print("0 <- Fuchsia")
print("1 <- Golden")
print("Labels:")
print(np.array(y))
```

### Wavy lines

```python
data, freq = wavy_lines(samples=3000, freq=1)
fig = plot_wavy(data,freq)
plt.show()
```

```python
data, freq = wavy_lines(samples=3000, freq=1)

x = np.array([i[0] for i in data]) 
y = np.array([i[1] for i in data]) #labels

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)

print("Train/Test datasets")
print(f"Train: {len(x_train)} samples")
print(f"Train: {len(x_test)} samples")
```

```python
#Labels
print("0 <- Bellow both curves")
print("1 <- Below f1 and above f2")
print("2 <- Above f1 and below f2")
print("3 <- Above both curves")
print(y)
```
